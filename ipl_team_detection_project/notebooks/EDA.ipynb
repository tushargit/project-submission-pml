```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('../master_csv.csv')

print(df.head())
print(df.shape)

labels = []

for col in df.columns[2:]:
    labels.extend(df[col].values)

sns.countplot(x=labels)
plt.title('Class Distribution')
plt.show()
```
