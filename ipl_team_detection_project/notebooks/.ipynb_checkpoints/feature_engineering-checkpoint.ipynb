```python
from src.utils import load_image
from src.grid_split import GridSplitter
from src.feature_extraction import FeatureExtractor

import matplotlib.pyplot as plt
import cv2

image_path = '../dataset/train/images/sample.jpg'

image = load_image(image_path)

splitter = GridSplitter()
extractor = FeatureExtractor()

cells = splitter.split_into_cells(image)

cell = cells[0]

features = extractor.extract_features(cell)

print(features.shape)

plt.imshow(cv2.cvtColor(cell, cv2.COLOR_BGR2RGB))
plt.show()
```
